# 0. Intro

`1` 같은 환경에 살아도 어떤 사람은 당뇨병에 걸리고 어떤 사람은 걸리지 않음.

`2` 식습관이나 운동 같은 환경 요인도 중요하지만, **태어날 떄부터 가지고 있는 유전적 차이**도 큰 역할을 한다.

`3` 그렇다면 다음과 같은 질문이 생김

> "유전적으로 어떤 차이가 질병 위험을 바꾸는가?" 

`4` 이 질문에 답하기 위한 연구 방법이 바로 GWAS(Genome-wide Association Study, 전장유전체 연관 분석)이다.

<details>
<summary> 용어 정리</summary>    

1. Genome Wide: 전장 유전체: 모든 유전체 위치에 대해서

2. Association Study = 연관분석: 관심을 가진 형질(Target phenotype)과 연관성을 갖는 유전적 위치를 찾는 것

</details>

# 1. GWAS란?

`1` GWAS는 유전자 전체의 변이를 조사해서, 특정 질병이나 형질(특성)과 **"통계적으로 연관된"** 변이를 찾아내는 연구방법이다.

`2` 즉, **"수십만-수백만 개의 유전변이를 한 번에 쭉 훑어보면서, 어느 변이가 병과 관련이 있는지 찾아내는 대규모 스캔"** 이라고 할 수 있음

# 2. 사전지식!

## DNA, Gene, Genome

`1` DNA: 생명의 설계도(유전정보) $\to$ A, T, G, C 네 가지 글자로 이루어진 긴 문자열.

`2` Gene(유전자): DNA중에서 단백질 만드는 방법이 적힌 구간, 약 2만 개가 있음.

`3` Genome(유전체): 한 개체가 지닌 DNA 정보 전체 총합, 약 32억 글자

Single Nucleotide Polymorphism
