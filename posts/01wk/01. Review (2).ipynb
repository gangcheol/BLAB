# 01. 연구배경 및 문제정의

## (1) 연구배경 및 핵심문제

`1` GWAS를 통해 발굴된 공통변이(Common Variant)들은 `Complex diseasses`의 유전력 중 일부만 설명할 수 있어. Missing Heritability 문제가 발생함.

`2` 이에 대한 대안으로 Next-Generation Sequencing(NGS) 기술을 활용해 희귀변이(Rare Variants)를 발굴하려는 연구가 활성화 되었음.

`3` 그러나 희귀변이는 표본 수가 극히 적어 `단일 변이 분석`(Single-Variant test)를 적용할 경우 통계적 검정력이 현저히 떨어짐.

`4` 이를 극복하기 위해 유전자나 특정 영역 단윋르로 변이들을 집계하여 검정하는 `Gene-level aggregation test`가 필수적으로 요구되었음.

## (2) 기존 Baseline 방법론들의 한계점

### a. Burden test(CAST, WST, CMC 등)의 한계

`1` 유전자 영역 내 모든 회귀변이가 `동일한 방향`(Unidirectional, 예: 모두 위험 변이)과 유사한 효과 크기를 가진다는 강한 `암묵적 가정`(Implicit Assumption)에 의존함.

`2` 그러나 실제 유전체 영역에는 효과가 없는 비원인 변이(Non-causal variants)나 보호 변이(Protective variants)가 혼재되어 있는 경우가 많으며,

`3` 이 경우 단순 누적 방식?은 상당한 노이즈를 유입시켜 분석 결과의 검정력을 대폭 저하시킴....

<details>
<summary>발췌 원문</summary>

> A limitation for all these burden tests is that they implicitly assume that all rare variants influence the phenotype in the same direction and with the same magnitude of effect... Hence, collapsing across all variants is likely to introduce substantial noise into the aggregated index, attenuate evidence for association, and result in power loss

</details>

### b. 초기 Non-Burden Test(C-alpha test 등)의 한계

`1` 변이 효과의 방향성에 구애받지 않는다는 장점이 있으나, 인구 집단 층화?(Population stratification) 제어를 위한 공변량(Corvariates, 예: Principal Components) 조정이 어려움.

`2` 또한 연관 불균형(Linkager Disequilibrium)이 존재하는 경우 `p-value` 계산시 수많은 'Permutation'에 의존하여 전장 유전체 수준(Genome-Wide scale)에서의 계산 비용이 비현실적으로 크게됨.

### c. 소규모 표본 연구에서 Large-Sample Asymptotics 한계

`1` 표준 SKAT는 대규모 표본에 기반한 점근적 분포(Asymptotic distribution)를 사용하여 p-value를 계산함.

`2` 하지만 희귀 질한 연구와 같이 케이스-컨트롤 표본 크기가 수십~수백 명 수준으로 작은 이분형 형질(Dichotomous traits) 연구에서는 제 1종 오류가 지나치게 보수적(Conservative)으로 형성되어 진정한? 연관 신호를 놓치게 됨.

### d. Public Controls 활용 시의 Technical Artifacts

`1` 희귀질병 연구시 자체 컨트롤 코호트?를 추가 시퀀싱 하는 cost는 사실 매우 부담스러움.

`2` gnomAD/ExAC과 같은 대규모 공공 데이터를 컨트롤로 활용할 경우, 케이스와 컨트롤 간 시퀀싱 플랫폼 차이, Joint variant calling 부재, Read depth 및 Quality score의 계통적 불일치로 인해,

`3` 유의성이 인위적으로 부풀려지는(Artifactual inflation) 치명적인 문제가 발생함.

# 02. Method & Key Results

## (1) 방법론 발전 과정

- 이거 이미지로 그려달라고 하기

### step1. SKAT의 등장: 단방향 가정 탈피

#### a. 발전 배경 및 모형 수립

`1` 기존 Burden test는 변이들을 하나로 합치는 과정에서 Protective 변이와 Deleterious 변이가 서로의 효과를 상쇄시키거나,

`2` 비원인 변이가 노이즈로 작용해 검정력이 붕괴되었음.

`3` 이를 극복하기 위해 개별 변의의 효과 회귀계수 $\beta_j$가 평균 0, 분산 $w_j^2 \tau$인 임의의 분포를 따른다고 가정한 **SKAT (Sequence Kernel Association Test)** 을 제안됨.

<details>
<summary>원문</summary>

> "SKAT uses a multiple regression model to directly regress the phenotype on genetic variants in a region and on covariates, and so allows different variants to have different directions and magnitude of effects, including no effects; SKAT also avoids selection of thresholds.

</details>

- 검정통계량 $Q$

$$Q = (y - \hat{\mu})'K (y - \hat{\mu}) = \sum_{j=1}^p w_j^2 S_j^2$$

>  여기서 $K = G W W G'$는 커널 행렬이며, $\sqrt{w_j} = \text{Beta}(\text{MAF}_j, 1, 25)$ 가중치를 적용하여 희귀변이의 기여도를 끌어올림.

#### b. 검정 결과

`1` **시뮬레이션 검정력 비교**: 이거 다시 결과 봐야함.....

- Causal variants의 효과 방향이 양방향(50% positive / 50% negative)으로 섞인 조건에서,
- 기존 Burden tests (WST, CAST, Counting)는 검정력이 0.10.2 이하로 완전히 붕괴된 반면,
- SKAT은 $n=5,000$에서 0.60.7 이상의 높은 검정력을 유지했습니다

`2` 실제 데이터 적용 결과: Table 4

- ANGPTL3, ANGPTL4, ANGPTL5 유전자 내 희귀변이와 연속형 혈청 Triglyceride 수치 연관성 분석에서, SKAT은 $p = 9.5 \times 10^{-5}$를 기록하여,

-  CAST ($1.9 \times 10^{-3}$) 및 WST ($2.3 \times 10^{-4}$) 대비 월등히 우수한 유의성을 검출함.

### step2. SKAT-O: 유전적 구조에 구애받지 않는 최적 파워 확보

#### a. 발전 배경 및 모형 수립

`1` SKAT은 변이 효과가 섞여 있을 때 우수하지만, 유전자 내 원인변이 비율이 높고(예: 50%)

`2` 모두 동일한 방향일 때에는 오히려 Burden test가 SKAT보다 더 우수한 검정력을 보이는 딜레마가 존재했음.

`3` 이에 따라  두 검정통계량을 회귀계수 간 상관계수 파라미터 $rho$로 가중 결합한 **SKAT-O (Optimal Unified Test)** 가 제안되었음.

$$Q_\rho = \rho Q_B + (1-\rho) Q_S, \quad Q_{\text{optimal}} = \min_{0 \le \rho \le 1} p_\rho$$

-  $\rho=1$이면 Burden test로, $\rho=0$이면 SKAT으로 수렴하며, 데이터로부터 파워를 극대화하는 최적의 $\rho$를 적응적으로 선택

<details>
<summary>원문</summary>

> &gt; *"Burden tests are more powerful when most variants in a region are causal and the effects are in the same direction, whereas SKAT is more powerful when a large fraction of the variants in a region are noncausal or the effects of causal variants are in different directions. The proposed unified test maintains the power in both scenarios."*

</details>

#### b. 검정 결과: 강건성 확보

Figure 1, 2, 3

`1` *Causal variants = 10% (희소 원인변이)*:

- SKAT과 SKAT-O가 Burden test 대비 압도적인 파워를 보임

`2` *Causal variants = 50% &amp; Unidirectional (높은 원인변이 비율)*:

- Burden test가 SKAT보다 우수한 구간에서,
- SKAT-O는 높은 $\rho$ 값($\rho \approx 1$)을 자동으로 선택하여 Burden test에 필적하는 최상의 파워를 보임

`3` *Mixed Effects (20% 또는 50% protective variants)*:

- Burden test는 효과 상쇄로 파워가 반토막 이하로 급감했으나, SKAT-O는 SKAT 수준 이상의 높은 검정력을 강건하게 유지함

### step3. Small-Sample Adjustment

#### a. 발전 배경 및 모형 수립

`1` 소규모 이분형 케이스-컨트롤 시퀀싱 연구($n < 1,000$)에서는 희소 데이터 특성으로 인해 표준 SKAT/SKAT-O의 점근적 $p$-value가 지나치게 보수적(Deflated)으로 형성되어 진정한 연관 신호를 놓치는 문제가 발생했음.

`2` 이에 따라 소표본 분산 및 귀무가설 하에서 Bootstrap Resampling(10,000회)을 통한 Sample Kurtosis  $\hat{\gamma}$ 추정을 도입하여 `Small-Sample Adjustment`를 완성함.

<details>
<summary> 원문 </summary>

> "The unadjusted SKAT and unified test (SKAT-O) had substantially deflated type I error rates for small sample sizes... Both the proposed small-sample-adjusted SKAT and the unified test (adjusted SKAT and adjusted SKAT-O) performed much better than their unadjusted counterparts in small samples."

</details>

#### b. 검정 결과

`1` 제 1종 오류 정상화(table 1)

- Exome-wide 수준($\alpha = 2.5 \times 10^{-6}$, $n=200$)에서, 미보정 SKAT의 Type I error는 $1.00 \times 10^{-7}$, SKAT-O는 $3.00 \times 10^{-7}$로 심각하게 왜곡되었으나,

- Adjusted SKAT ($3.20 \times 10^{-6}$) 및 **Adjusted SKAT-O ($2.50 \times 10^{-6}$)**는 목표 유의수준을 완벽하게 통제함

`2` 실제 데이터 적용 결과(Figure 4)

- 88명의 급성 폐손상 환자 데이터에서 기존 미보정 SKAT/SKAT-O의 Q-Q plot은 아래로 강하게 왜곡(deflated)되었으나,

- Adjusted SKAT-O Q-Q plot은 $45^\circ$ 대각선에 일치했습니다76more_horiz. 이를 통해 급성 폐손상 원인 유전자인 MYLK 유전자가 상위 유의성(전체 4위)으로 검출됨.

### step4. Public Controls(gnomAD/ExAC) 활용 및 TRAPD 아티팩트 보정

#### a. 발전 배경

`1` 희귀질환 연구 시 자체 대조군을 시퀀싱하는 비용을 절감하기 위해 gnomAD/ExAC 등 공공 데이터베이스를 컨트롤로 활용하려는 시도가 이루어졌음.

`2` 그러나 시퀀싱 플랫폼 차이 및 별도 Variant calling으로 인한 기술적 아티팩트(False positive)가 발생함...

`3` 이에 따라 해당 논문에서는  Guo et al. (2018)은 TRAPD 프레임워크를 개발하고,

`4` Read-Depth 필터링($>10\times$ depth in $\ge 90\%$ samples), Quality-by-Depth (QD) 정량 필터링(gnomAD 상위 95%, Case 상위 85%),ㅡ

`5` 그리고 Synonymous variant 캘리브레이션 지표 **$\lambda_{D95} \approx 1.00$**을 제안했음.

<details>
<summary> 원문 </summary>

> *"We leveraged presumably benign synonymous variants to calibrate our approach. Through iterative analyses, we systematically addressed and overcame various sources of artifact that can arise when using public control data."*

</details>

#### b. 검정 결과

# 03. 결론 및 시사점

## (1) 종합 결론

`1` 유전체 희귀변이 연관성 분석 방법론은 단일 변이 검정의 파워 부족을 해결하기 위해 **SKAT**과 같은 분산 성분 검정으로 발전하였고,

`2` 나아가 Burden test와의 파워 트레이드오프를 데이터 적응적으로 해소하는 **SKAT-O** 프레임워크로 진화함.

`3` 더불어 **Small-Sample Adjustment**를 통해 소규모 코호트 연구에서도 수학적으로 정확한 Type I error 통제와 파워 향상을 이뤄냈으며,

`4` **TRAPD 프레임워크**를 통해 gnomAD와 같은 공공 데이터베이스를 컨트롤로 안전하게 결합할 수 있는 정량 필터링 기법이 정립되었음.

## (2) 시사점

`1` **유전적 구조에 대한 높은 강건성 (Robustness)**:

- 원인 변이의 비율, 변이 효과의 방향성, LD 구조, 케이스 표본 크기에 관계없이 최적의 통계적 검정력을 안정적으로 확보함.

`2` **연구 비용 및 자원 절감 (Cost Efficiency)**

- 희귀 질환 연구에서 자체 대조군 코호트를 대규모로 추가 시퀀싱하지 않고도 gnomAD/ExAC 공공 데이터를 대조군으로 활용할 수 있게 되어 연구 자원이 제한된 환경에서도 희귀질환 원인 유전자 발굴 연구를 대폭 촉진

`3` **정밀 의료 및 미지의 유전력 규명 (Translational Discovery)**:

- *MYLK*, *TYRO3* 등의 신규 원인 유전자 발굴 사례에서 입증되었듯, 정밀 의료(Precision Medicine) 구현을 위한 희귀/복잡 질환의 Missing Heritability 해소에 결정적인 기여
