# 01. 연구배경 및 문제정의

## (1) 연구배경 및 핵심문제

`1` GWAS를 통해 발굴된 공통변이(Common Variant)들은 `Complex diseasses`의 유전력 중 일부만 설명할 수 있어. Missing Heritability 문제가 발생함.

`2` 이에 대한 대안으로 Next-Generation Sequencing(NGS) 기술을 활용해 희귀변이(Rare Variants)를 발굴하려는 연구가 활성화 되었음.

`3` 그러나 희귀변이는 표본 수가 극히 적어 `단일 변이 분석`(Single-Variant test)를 적용할 경우 통계적 검정력이 현저히 떨어짐.

`4` 이를 극복하기 위해 유전자나 특정 영역 단윋르로 변이들을 집계하여 검정하는 `Gene-level aggregation test`가 필수적으로 요구되었음.
