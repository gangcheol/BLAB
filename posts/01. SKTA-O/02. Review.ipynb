# Burden vs SKAT

$$Q_{\text{Burden}} = \left [ \sum_{j=1}^{m} w_jS_j \right ]^2 \quad \text{vs} \quad Q_{\text{SKAT}} =\sum_{j=1}^m w_j^2 S_j^2$$

|예시| $w = 1, \,\,S = [1,2,3]$| $w = 1, \,\,S =[4,5,-6]$|
|:---:|:--------:|:--------:|
|$Q_{\text{Burden}}$|36|9|
|$Q_{\text{SKAT}}$|14|77|

# SKAT-O

`1` 위 같은 한계점을 보완하기 위해, **때로는 Burden처럼 때로는 SKAT처럼** 작동하고 싶은게 아이디어의 출발점이다.

`2` 아래의 `식1`을 `식2`로 유도하는 과정에서 위가 어떻게 작동할 수 있는지 원리를 깨달을 수 있다.

- Euqation 1

$$Q_\rho = \rho Q_B + (1 - \rho) Q_S \quad (0 \le \rho \le 1)$$

- Equation 2

$$Q_\rho = (\mathbf{y} - \hat{\boldsymbol{\pi}}_0)^T \mathbf{K}_\rho (\mathbf{y} - \hat{\boldsymbol{\pi}}_0)$$

## Step1. 기본 기호 정의

`1` 잔차 벡터: $\mathbf{r} = \mathbf{y} - \hat{\boldsymbol{\pi}}_0$

`2` 변이별 가중치 대각 행렬: $\mathbf{W} = \text{diag}(w_1, w_2, \dots, w_m)$

`3` 변이 스코어 행렬: $\mathbf{S} = \mathbf{G}^T (\mathbf{y} - \hat{\boldsymbol{\pi}}_0) = \mathbf{G}^T \mathbf{r}$

-  $S_j = \sum_{i=1}^n g_{ij}(y_i - \hat{\pi}_{0,i})$) 

## Step2. 각 Q를 행렬 표현

`1` SKAT

$$Q_{\text{SKAT}}= \sum_{j=1}^m w_j^2 S_j^2  = \mathbf{S}^T \mathbf{W} \mathbf{I}_m \mathbf{W} \mathbf{S}$$

`2` Buredn

$$Q_{\text{Burden}} = \left( \mathbf{1}^T \mathbf{W}\mathbf{S} \right)^2 = \left( \mathbf{1}^T \mathbf{W}\mathbf{S} \right)^T \left( \mathbf{1}^T \mathbf{W}\mathbf{S} \right) = \mathbf{S}^T \mathbf{W} (\mathbf{1}\mathbf{1}^T) \mathbf{W} \mathbf{S}$$

## Step3. 행렬 묶기

$$
\begin{aligned}
Q_\rho &= (1 - \rho) Q_S + \rho Q_B \\ \\
&= (1 - \rho) \left( \mathbf{S}^T \mathbf{W} \mathbf{I} \mathbf{W} \mathbf{S} \right) + \rho \left( \mathbf{S}^T \mathbf{W} (\mathbf{1}\mathbf{1}^T) \mathbf{W} \mathbf{S} \right) \\ \\
&= \mathbf{S}^T \mathbf{W} \left[ \mathbf{(1 - \rho) \mathbf{I} + \rho \mathbf{1}\mathbf{1}^T} \right] \mathbf{W} \mathbf{S} \\ \\
&= \mathbf{S}^T \mathbf{W}\mathbf{R}_\rho \mathbf{W} \mathbf{S}
\end{aligned}
$$

## Step4. 잔차 벡터와 커널 행렬로 전환

$$
\begin{aligned}
Q_\rho &= \mathbf{S}^T \mathbf{W} \mathbf{R}_\rho \mathbf{W} \mathbf{S} \\ \\
&= \left( \mathbf{G}^T (\mathbf{y} - \hat{\boldsymbol{\pi}}_0) \right)^T \mathbf{W} \mathbf{R}_\rho \mathbf{W} \left( \mathbf{G}^T (\mathbf{y} - \hat{\boldsymbol{\pi}}_0) \right) \\ \\
&= (\mathbf{y} - \hat{\boldsymbol{\pi}}_0)^T \underbrace{\left( \mathbf{G} \mathbf{W} \mathbf{R}_\rho \mathbf{W} \mathbf{G}^T \right)}_{\mathbf{K}_\rho} (\mathbf{y} - \hat{\boldsymbol{\pi}}_0) \\
&= (\mathbf{y} - \hat{\boldsymbol{\pi}}_0)^T {\mathbf{K}_\rho} (\mathbf{y} - \hat{\boldsymbol{\pi}}_0) 
\end{aligned}
$$

## Step5. 커널 행렬 뜯어보기

커널 행렬 $\mathbf{K}_\rho$에서 $\mathbf{R}_\rho = (1-\rho)\mathbf{I} + \rho\mathbf{1}\mathbf{1}^T$를 다시 넣어서 풀면

$$
\begin{aligned}
\mathbf{K}_\rho &= \mathbf{G} \mathbf{W} \left[ (1 - \rho)\mathbf{I} + \rho \mathbf{1}\mathbf{1}^T \right] \mathbf{W} \mathbf{G}^T \\ \\ 
&= (1 - \rho) \underbrace{\mathbf{G} \mathbf{W} \mathbf{W} \mathbf{G}^T}_{\mathbf{K}_{\text{SKAT}}} + \rho \underbrace{\mathbf{G} \mathbf{W} \mathbf{1}\mathbf{1}^T \mathbf{W} \mathbf{G}^T}_{\mathbf{K}_{\text{Burden}}}
\end{aligned}
$$

- 즉, 위 행렬 유도를 통해 독립성을 가정하면($\rho = 0$), SKAT처럼,

- 즉, 위 행렬 유도를 통해 독립성을 가정을 배제하면 ($\rho \neq 0$), Burden처럼 작동하는 것이다.
