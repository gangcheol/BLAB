# Burden

여러 희귀변이들을 **단 하나의 Burden Score($Q_{\text{Burden}}$)로** 압축하여 분석하는 검정법

## 모형 & 가설

$$H_0: \beta_c = 0 \quad \text{vs} \quad H_1: \beta_c \neq 0$$


$$\begin{aligned}\text{logit} (\pi_{i}) = \log \left(\frac{\pi_i}{1-\pi_i} \right) &= \alpha_0 + \boldsymbol{\mathrm{X}}_i^{\top}\boldsymbol{\alpha}
                + \beta_c\left (\sum_{j=1}^{m} w_j g_{ij}\right) \\ 
                &= \alpha_0 + \boldsymbol{\mathrm{X}}_i^{\top}\boldsymbol{\alpha} + \beta_c C_{i} \end{aligned}$$


- $\boldsymbol{\alpha}$: 공변량(나이, 성별, 주성분 등)의 효과 계수 $\to$ `fixed effect`

- $\beta_c$: 공통 효과 크기 $\to$ `fixed effect`

- $w_j$: 변이 $j$의 `MAF`기반 사전 가중치
    - $w_j = \text{Beta}(\text{MAF}_j;, 1, 25)$

## 검정통계량

$$Q_{\text{Burden}} = \left [\sum_{i=1}^{n}(y_i-\hat {\pi}_i) \left (\sum_{j=1}^{m}w_jg_{ij}\right) \right ]^{2} = \sum_{j=1}^{m} w_js_j^{2}$$

- $y_i- \hat{\pi}_i$: 귀무모형 하에서의 잔차
- $g_{ij}:$ $i$번째 환자의 $j$번째 변이에 대한 Genotype ($0,1,2$)
- $w_j$: 변이 $j$의 가중치 (예: $\text{Beta}(\text{MAF}\_j; 1, 25)$)
- $S_j = \sum_{i=1}^n g_{ij}(y_i - \hat{\pi}_{i})$: 개별 변이 $j$의 스코어
- **핵심 특징**: 모든 변이의 스코어를 **먼저 합산 후 제곱**

## Key point

`1`. `Burden Test`는 변이 개수를 사전에 가중합산하여 단일 변수 $C_i$로 만든 후 회귀모형에 적합.

`2` 개별 $C_i$는 다음과 같은 행렬 매트릭스 $\boldsymbol{\mathrm{C}}$의 $i$번째 원소.

$$\begin{aligned}\text{logit} (\pi_{i}) = \log \left(\frac{\pi_i}{1-\pi_i} \right) &= \alpha_0 + \boldsymbol{\mathrm{X}}_i^{\top}\boldsymbol{\alpha}
                + \beta_c\left (\sum_{j=1}^{m} w_j g_{ij}\right) \\ \\ 
                &= \alpha_0 + \boldsymbol{\mathrm{X}}_i^{\top}\boldsymbol{\alpha} + \beta_c C_{i} \end{aligned}$$

$$
\mathbf{C} = \mathbf{G} \mathbf{w} = \begin{bmatrix} g_{11} & g_{12} & \dots & g_{1m} \\ g_{21} & g_{22} & \dots & g_{2m} \\ \vdots & \vdots & \ddots & \vdots \\ g_{n1} & g_{n2} & \dots & g_{nm} \end{bmatrix}_{n \times m} 
\begin{bmatrix} w_1 \\ w_2 \\ \vdots \\ w_m \end{bmatrix}_{m \times 1} = \begin{bmatrix} C_1 \\ C_2 \\ \vdots \\ C_n \end{bmatrix}_{n \times 1}
$$

## Why GLM?

`1`. `Burden Test`는 변이 개수를 사전에 가중합산하여 단일 변수 $C_i$로 만든 후 회귀모형에 적합.

`2` 개별 $C_i$는 다음과 같은 행렬 매트릭스 $\boldsymbol{\mathrm{C}}$의 $i$번째 원소.

$$\begin{aligned}\text{logit} (\pi_{i}) = \log \left(\frac{\pi_i}{1-\pi_i} \right) &= \alpha_0 + \boldsymbol{\mathrm{X}}_i^{\top}\boldsymbol{\alpha}
                + \beta_c\left (\sum_{j=1}^{m} w_j g_{ij}\right) \\ \\ 
                &= \alpha_0 + \boldsymbol{\mathrm{X}}_i^{\top}\boldsymbol{\alpha} + \beta_c C_{i} \end{aligned}$$

-  반응변수 분포 가정 $\to y_i \sim \text{Ber}(\pi_i)$
-  검정 모수 $\beta_c$가 확률변수가 아닌 고정된 단일 상수 (`Fixed effect`)
-  $m$개의 변이를 단일 변수 $C_i$로 묶어 모형을 적합 ($df = 1$)
-  연결함수 $g(\cdot)$를 활용하여 이분형일 경우, 로짓으로 적합.

## 한계점

1. 상쇄효과: 유전자 내 변이 신호가 혼재되어 있으면, 검정력이 낮아짐.

2. 또한, 모든 변이가 **같은 방향과 같은 크기**($\beta_c$)를 갖는다는 비현실적인 가정을 전제로 하여 모형을 적합함.
    - $w_j$를 모두 1이라 가정

|예시| S = $[1,2,3]$| S = $=[1,2,-3]$|
|:---:|:---:|:---:|
|$Q_{\text{Burden}}$|36|0|
|$Q_{\text{SKAT}}$|?|

# SKAT

여러 희귀변이들을 단 하나의 점수(Burden Score)로 집계하지 않고, 각 변이 계수($\beta_j$)가 임의의 확률분포를 따른다고 가정함. 

## 모형 & 가설

$$H_0: \tau = 0 \quad \text{vs} \quad H_1: \text{not } H_0$$

$$\text{logit}(\pi_i) = \alpha_0 + \mathbf{X}_i^T \boldsymbol{\alpha} + \sum_{j=1}^m \beta_j g_{ij}$$

- $\boldsymbol{\alpha}$: 공변량(나이, 성별, 주성분 등)의 효과 계수 $\to$ `fixed effect`

-  **임의 효과(Random Effect) 가정**
    - 개별 $\beta_j$를 상수로 추정하지 않고 확률분포로 설정: $\beta_j \sim_{iid} (0, \tau w_j^2)$
    - $\tau$: 변이 집합 전체의 효과 크기를 제어하는 **공통 분산 모수**

## 검정통계량

$$Q_{\text{SKAT}} = (\mathbf{y} - \hat{\boldsymbol{\pi}}_0)^T \mathbf{K} (\mathbf{y} - \hat{\boldsymbol{\pi}}_0) = \sum_{j=1}^m w_j^2 S_j^2$$

  * $\mathbf{K} = \mathbf{G}\mathbf{W}\mathbf{W}\mathbf{G}^T$: 환자 간 유전적 거리/유사도 커널 행렬 ($n \times n$)
  * $S_j = \sum_{i=1}^n g_{ij}(y_i - \hat{\pi}_{0,i})$: 개별 변이의 스코어
  * $S_j^2$: 개별 스코어를 **먼저 제곱** $\rightarrow$ 부호 상쇄 문제 원천 차단

* **귀무분포 (카이제곱 혼합 분포)**: $Q_S \sim \sum_{j=1}^m \lambda_j \chi^2_{1, j}$
  * $\lambda_j$: 정보-커널 결합 행렬의 고유값(Eigenvalue)

## SKAT 통계량 유도 과정

1. **커널 행렬 분해**
$$\mathbf{K} = \mathbf{G} \mathbf{W} \mathbf{W} \mathbf{G}^T$$

2. **이차형식 대입 및 벡터 묶기**
$$Q_S = \left[ (\mathbf{y} - \hat{\boldsymbol{\pi}}_0)^T \mathbf{G} \mathbf{W} \right] \left[ \mathbf{W} \mathbf{G}^T (\mathbf{y} - \hat{\boldsymbol{\pi}}_0) \right] = \mathbf{v}^T \mathbf{v}$$

3. **스코어 벡터 $\mathbf{S}$ 치환 ($\mathbf{S} = \mathbf{G}^T (\mathbf{y} - \hat{\boldsymbol{\pi}}_0)$)**
$$\mathbf{v} = \mathbf{W} \mathbf{S} = [w_1 S_1, w_2 S_2, \dots, w_m S_m]^T$$

4. **벡터 내적 계산**
$$Q_S = \mathbf{v}^T \mathbf{v} = \sum_{j=1}^m (w_j S_j)^2 = \sum_{j=1}^m w_j^2 S_j^2$$

## SKAT이 GLMM인 이유와 한계점

-  반응변수 분포 가정 $\to y_i \sim \text{Ber}(\pi_i)$

* 공변량 $\boldsymbol{\alpha}$ = 고정 효과 (`Fixed Effect`)

* 변이 효과 $\boldsymbol{\beta}$ = 임의 효과 (`Random Effect`, $\text{Var}=\tau w_j^2$)

* 임의효과의 분산 성분 검정(Score test)을 수행

* 유효자유도 제어를 통해 변이의 수가 많아도 $df$ 폭발 방어

|예시| S = $[1,2,3]$| S = $=[1,2,-3]$|
|:---:|:---:|:---:|
|$Q_{\text{Burden}}$|36|0|
|$Q_{\text{SKAT}}$|?|

## 데이터 생성

```r
library(SKAT)
set.seed(2026)

# 1. 데이터 생성 (N = 1000, 희귀 변이 M = 30)
N <- 1000; M <- 30
maf <- runif(M, 0.001, 0.01)
Z <- sapply(maf, function(p) rbinom(N, 2, p)) ## 지노타입 행렬
X <- rnorm(N) ## X 공변량

# 가중치 설정: SKAT 기본 가중치인 Beta(MAF, 1, 25) 동일 적용
maf_est <- colMeans(Z) / 2
w <- dbeta(maf_est, 1, 25)
W <- diag(w)

# 시나리오 1 (단방향: 인과 변이 14개 모두 양의 효과)
beta_uni <- c(rep(0.7, 14), rep(0, 16))
y_uni <- 0.5 * X + Z %*% beta_uni + rnorm(N)

# 시나리오 2 (양방향: 인과 변이 7개는 +0.7, 7개는 -0.7)
beta_bi <- c(rep(0.7, 7), rep(-0.7, 7), rep(0, 16))
y_bi <- 0.5 * X + Z %*% beta_bi + rnorm(N)

# 2. Score 통계량 및 유효 자유도 계산 함수
analyze_score_df <- function(y, Z, X, w) {
  # 귀무모형 적합
  null_fit <- lm(y ~ X)
  r <- residuals(null_fit)
  sigma2 <- sum(r^2) / null_fit$df.residual
  
  # 투영 행렬 P0 구성 (연속형 기준)
  X_tilde <- model.matrix(null_fit)
  P0 <- diag(nrow(Z)) - X_tilde %*% solve(t(X_tilde) %*% X_tilde) %*% t(X_tilde)
  
  # 개별 변이의 Score 벡터 S (SKAT 패키지의 1/2 스케일링 컨벤션 반영 가능)
  # 여기서는 표준 정의 S = G^T r / sigma2 사용
  S <- as.vector(t(Z) %*% r) / sigma2
  
  # [Burden] 가중 합산 Score 및 통계량
  S_burden_sum <- sum(w * S)                 # 부호 상쇄가 일어나는 지점!
  Q_burden <- S_burden_sum^2                 # 비표준화 Burden Q
  var_burden <- as.numeric(t(w) %*% t(Z) %*% P0 %*% Z %*% w / sigma2)
  chisq_burden <- Q_burden / var_burden      # ~ Chi-sq(df = 1)
  df_burden <- 1.0
  
  # [SKAT] 가중 제곱합 통계량 및 고유값 기반 유효 자유도
  Q_skat <- sum((w * S)^2)                   # 제곱 후 합산하므로 상쇄 안 됨!
  K <- diag(w) %*% t(Z) %*% P0 %*% Z %*% diag(w) / sigma2
  lambdas <- eigen(K, symmetric = TRUE, only.values = TRUE)$values
  lambdas <- lambdas[lambdas > 1e-8]
  
  # Satterthwaite 유효 자유도 (Effective DF)
  df_skat_eff <- (sum(lambdas)^2) / sum(lambdas^2)
  # 귀무가설 하에서 Q_skat의 기댓값 E[Q] = sum(lambdas)
  expected_Q_skat <- sum(lambdas)
  
  return(c(
    `Sum(w*S)`       = S_burden_sum,
    `Q_Burden`       = Q_burden,
    `ChiSq_Burden`   = chisq_burden,
    `df_Burden`      = df_burden,
    `Q_SKAT`         = Q_skat,
    `E[Q_SKAT]_Null` = expected_Q_skat,
    `Eff_df_SKAT`    = df_skat_eff
  ))
}

# 3. 두 시나리오 비교 테이블 출력
comp_table <- rbind(
  Unidirectional = analyze_score_df(y_uni, Z, X, w),
  Bidirectional  = analyze_score_df(y_bi, Z, X, w)
)

knitr::kable(round(comp_table, 2), caption = "효과의 방향성에 따른 Score 통계량과 자유도 비교")
```

```r
# 1. Install and load the SKAT package
# install.packages("SKAT")
library(SKAT)

# 2. Load the built-in example dataset
data(SKAT.example)

# The dataset contains:
# Z: Genotype matrix (2000 individuals, 67 SNPs)
# X: Covariate matrix (2 covariates)
# y.c: Continuous phenotype vector
# y.b: Binary phenotype vector

Z <- SKAT.example$Z
X <- SKAT.example$X
y.c <- SKAT.example$y.c
y.b <- SKAT.example$y.b

# 3. Fit the null model (adjusting for covariates X, continuous outcome)
obj <- SKAT_Null_Model(y.c ~ X, out_type = "C")

# 4. Run SKAT association test
out <- SKAT(Z, obj)

# 5. View results
print(out)
```
