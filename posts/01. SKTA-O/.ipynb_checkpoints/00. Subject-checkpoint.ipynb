# 1. 검정법 비교(Wald, LRT, SCORE)

## Likelihood Ratio Test (LRT):

`1` 귀무가설 하에서의 최대 우도값 $L(\hat{\Omega}_0) = \max_{\boldsymbol{\theta} \in \Omega_0} L(\boldsymbol{\theta})$와 전체 모수 공간에서의 최대 우도값 $L(\hat{\Omega}) = \max_{\boldsymbol{\theta} \in \Omega} L(\boldsymbol{\theta})$의 비율로 검정통계량 $\lambda$를 정의

$$\lambda = \frac{L(\hat{\Omega}_0)}{L(\hat{\Omega})} = \frac{\max_{\boldsymbol{\theta} \in \Omega_0} L(\boldsymbol{\theta})}{\max_{\boldsymbol{\theta} \in \Omega} L(\boldsymbol{\theta})} \quad (0 \le \lambda \le 1) \quad$$

`2` 점근적 성질 (Wackerly Theorem 10.2): 표본 크기 $n$이 충분히 클 때, 정규성 조건(Regularity conditions) 하에서 $-2 \ln(\lambda)$는 점근적으로 카이제곱($\chi^2$) 분포를 따름.

## Wald Test (대표본 Z/카이스퀘어 검정)

`1` 대립가설 하에서 구한 최대우도추정량(MLE) $\hat{\theta}$와 귀무가설 하의 모수값 $\theta_0$ 간의 거리를 추정량의 표준오차 $\sigma_{\hat{\theta}}$[또는 피셔 정보 행렬 $I(\hat{\theta})$]로 나눈 통계량을 활용

$$Z = \frac{\hat{\theta} - \theta_0}{\sigma_{\hat{\theta}}} \sim N(0, 1) \quad \Longrightarrow \quad W = Z^2 = \frac{(\hat{\theta} - \theta_0)^2}{\text{Var}(\hat{\theta})} \sim \chi^2_1 \quad$$

## **Rao's Score Test (스코어 검정)**

`1` 로그우도함수 $l(\boldsymbol{\theta})$의 1차 미분인 스코어 함수(Score Function) $U(\boldsymbol{\theta}) = \frac{\partial l(\boldsymbol{\theta})}{\partial \boldsymbol{\theta}}$와 2차 미분 기대값인,

`2` 피셔 정보 행렬(Fisher Information) $I(\boldsymbol{\theta}) = -E\left[\frac{\partial^2 l(\boldsymbol{\theta})}{\partial \boldsymbol{\theta} \partial \boldsymbol{\theta}^T}\right]$을 바탕으로, 귀무가설 하 $\boldsymbol{\theta}_0$에서 평가

$$S = U(\boldsymbol{\theta}_0)^T \left[ I(\boldsymbol{\theta}_0) \right]^{-1} U(\boldsymbol{\theta}_0) \sim \chi^2_p$$

`3` 대립가설에서의 모수 추정치 $\hat{\boldsymbol{\theta}}$를 구하지 않고, 오직 귀무가설 지점 $\boldsymbol{\theta}_0$ 하에서 우도함수의 기울기(Slope, $U$)와 곡률(Information, $I$)만을 측정

### 비교 요약표

| 비교 항목 | Wald Test | Likelihood Ratio Test (LRT) | Rao's Score Test |
| :---: | :---: | :---: | :---: |
| **검정통계량 수식** | $W = \frac{(\hat{\theta} - \theta_0)^2}{\text{Var}(\hat{\theta})} \sim \chi^2_1$ | $\Lambda = -2 \ln \left[ \frac{L(\hat{\Omega}_0)}{L(\hat{\Omega})} \right] \sim \chi^2_{\nu}$ | $S = U(\boldsymbol{\theta}_0)^T \left[ I(\boldsymbol{\theta}_0) \right]^{-1} U(\boldsymbol{\theta}_0) \sim \chi^2_p$ |
| **수식 평가 위치** | $H_1$ 하의 MLE ($\hat{\theta}$) | $H_0$ ($\hat{\Omega}_0$) 및 $H_1$ ($\hat{\Omega}$) 모두 | **오직 $H_0$ ($\boldsymbol{\theta}_0$) 하에서만** |
| **통계량의 수리적 의미** | 모수 추정치 거리 $(\hat{\theta} - \theta_0)$ | 두 모수 공간의 우도비 $\frac{L(\hat{\Omega}_0)}{L(\hat{\Omega})}$ | $H_0$ 지점에서의 우도 기울기 $U(\boldsymbol{\theta}_0) = \left. \frac{\partial \ln L}{\partial \boldsymbol{\theta}} \right|_{\boldsymbol{\theta}_0}$ |
| **$H_1$ 하의 MLE 추정 필요성** | **필수** (MLE 수치 구함) | **필수** (MLE 수치 구함) | **불필요** ($H_0$ 수식만 활용) |
| **다중 검정 시 연산 구조** | $M$번의 수치 최적화 반복 | $M$번의 수치 최적화 반복 | **Null Model 1회 적합 + $M$번의 단순 내적** |

## Score Test 사용 시 연산량이 급감하는 원리

`1` 검정통계량 $S$의 평가 위치가 오직 귀무가설 지점 $\boldsymbol{\beta}_m = \mathbf{0}$에 고정되어 있으므로 대립모형에서의 모수 추정 과정을 완전히 소거.

`2` 관심 변수 $\boldsymbol{\beta}_m$이 소거된 귀무모형은 모든 $M$개 검정에서 동일한 성입모수 $\boldsymbol{\gamma}$만을 포함(우리 연구에서는 공변량만을 포함한다고 생각)

`3` 따라서 전체 분석 시작 시 **Null model을 단 1회만 적합(Fit Null Model Once)** 하여 기대값 $\hat{\mu}_{0i}$ 및 잔차 벡터 $r_i = y_i - \hat{\mu}_{0i}$를 구해둠.

$$\hat{\mu}_{0i} = \hat{\pi}_{0i} = P(y_i = 1 \mid \mathbf{X}_i) = \text{logit}^{-1}(\hat{\alpha}_0 + \mathbf{X}_i^T \hat{\boldsymbol{\alpha}})$$

- (유전자와 상관없이, 이 사람의 **나이, 성별, 인종 구조**만 보았을 때 질병에 걸릴 확률이 얼마인지)

`4` 이후 $M$개 가설을 검정할 때는 미리 구한 공통 잔차 벡터와 타겟 변수 간의 **단순 행렬 내적 연산(Matrix Inner Product)** 만 수행하므로,

`5` 반복 계산이 선형대수 곱셈으로 전환되어 연산량이 급감.

$$Q = (y - \hat{\mu})'K (y - \hat{\mu}) = \sum_{j=1}^p w_j^2 S_j^2$$

## **중심극한정리(CLT)와 카이스퀘어 분포로의 점근적 수렴**

`1` 중심극한정리에 의해 대규모 표본에서 스코어 벡터 $U(\boldsymbol{\theta}_0)$는 점근적으로 정규분포를 따름.

`2` 정규 확률변수의 제곱합은 카이스퀘어($\chi^2$) 분포를 형성하므로, 표본을 무작위로 섞는 Permutation 없이도 카이스퀘어 표(Table)나 수치 적분 공식을 통해 $p$-value(유의확률)를 즉시 산출 가능.

### summary

`1` $H_0$ 귀무가설($H_0$) 하에서 스코어 통계량(및 그 이차형식)이 이론적으로 카이제곱(혼합) 분포를 완벽히 따른다는 점을 이용해,

`2` 표본을 수천만 번 섞는 Permutation 시뮬레이션 없이 해석적 공식 대입만으로 exact $p$-value를 즉시 산출할 수 있다.

`3` 오직 귀무가설($H_0$) 지점에서만 평가하므로 대립모형($H_1$) 하의 수치 최적화(Newton-Raphson) 반복 계산을 완전히 회피하고,

`4` Null Model 1회 적합으로 얻은 공통 잔차를 재사용해 **단순 행렬 내적 연산만으로 연산 시간을** 단축시킴.

# 2. For logit

## 확률의 정의구역 $[0, 1]$ 보장 (Bounded Domain)

`1` 선형 확률 모형(LPM: $\pi_i = \mathbf{X}_i \boldsymbol{\beta}$)을 적용하면 예측된 $\hat{\pi}_i$가 $0$ 미만이거나 $1$을 초과하는 치명적 결함이 발생

`2` $\text{logit}(\pi) = \ln\left(\frac{\pi}{1-\pi}\right)$은 확률 구간 $(0, 1)$을 실수 전체 $(-\infty, \infty)$로 1:1 일대일 매핑해줌.

`3` 역변환(Sigmoid) $\pi_i = \frac{e^{\mathbf{X}_i \boldsymbol{\beta}}}{1 + e^{\mathbf{X}_i \boldsymbol{\beta}}}$을 통해 선형 결합 $\mathbf{X}_i \boldsymbol{\beta}$가 어떤 실수를 갖더라도 항상 $0 < \hat{\pi}_i < 1$ 범위 내의 유효한 확률값을 보장함.

`4` 예를 들어 일반 선형회귀로 질병 확률을 예측하면 위험 요인이 많아질 때 확률이 $120\%$가 되거나 $-15\%$가 되는 기괴한 문제가 발생합

`5` 반면 Logit 변환을 거치면 S자 곡선 형태가 되어 X값이 아무리 극단적이어도 예측 확률을 항상 $0\%$와 $100\%$ 사이($[0, 1]$)로 예쁘게 가두어줌.
